```python
import os
import base64
from PIL import Image
from openai import OpenAI   

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

## Part 1: Native Function / Tool Calling (Equivalent to Llama 3.1 Tool Routing)

```python
# Define a custom tool schema for OpenAI function calling
tools = [
    {
        "type": "function",
        "function": {
            "name": "get_current_weather",
            "description": "Get the current weather for a given city",
            "parameters": {
                "type": "object",
                "properties": {
                    "location": {
                        "type": "string",
                        "description": "The city and state, e.g. Stockholm, Sweden",
                    },
                    "format": {
                        "type": "string",
                        "enum": ["celsius", "fahrenheit"],
                        "description": "The temperature unit to use.",
                    },
                },
                "required": ["location", "format"],
            },
        },
    }
]

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Whats the weather like in Stockholm, Sweden in celsius?"},
    ],
    tools=cast(list[ChatCompletionToolUnionParam], tools),
    tool_choice="auto",
)
```
