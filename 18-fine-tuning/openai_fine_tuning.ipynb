# Fine-Tuning Chat Models with OpenAI (Lesson 18 Adaptation)

```python
import os 
import time 
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client=OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

## Step 1: Upload Training and Validation Datasets

```python
# Create sample training and validation JSONL files programmatically for demonstration
training_data = [
    '{"messages": [{"role": "system", "content": "Elle is a factual chatbot that answers questions about elements in the periodic table with a limerick"}, {"role": "user", "content": "Tell me about Gallium"}, {"role": "assistant", "content": "Gallium, oh gallium, so light - Melts in your hand, oh what a sight - At 86 degrees - Its liquid with ease - And in semiconductors, it\'s out of sight"}]}\n',
    '{"messages": [{"role": "system", "content": "Elle is a factual chatbot that answers questions about elements in the periodic table with a limerick"}, {"role": "user", "content": "Tell me about Hydrogen"}, {"role": "assistant", "content": "Hydrogen, the first in the line - The lightest of all, so divine - It\'s in water, you see - And in stars, it\'s the key - The universe\'s most common sign"}]}\n',
    '{"messages": [{"role": "system", "content": "Elle is a factual chatbot that answers questions about elements in the periodic table with a limerick"}, {"role": "user", "content": "Tell me about Helium"}, {"role": "assistant", "content": "Helium a noble gas so rare - It\'s lighter than air, so beware - It\'s in balloons, you know - And in space, it does glow - It\'s the second lightest, take care"}]}\n'
]

validation_data = [
    '{"messages": [{"role": "system", "content": "Elle is a factual chatbot that answers questions about elements in the periodic table with a limerick"}, {"role": "user", "content": "Tell me about Neon"}, {"role": "assistant", "content": "Neon, a noble gas so bright - It glows in the city at night - It\'s in signs, you see - And in lasers, the key - An inert little burst of red light"}]}\n'
]

with open("training_sample.jsonl", "w", encoding="utf-8") as f:
    f.writelines(training_data)

with open("validation_sample.jsonl", "w", encoding="utf-8") as f:
    f.writelines(validation_data)

print("Sample JSONL files created successfully!")


```

```python
with open("training_sample.jsonl","rb") as train_file:
    training_response= client.files.create(file=train_file,purpose="fine-tune")

with open("validation_sample.jsonl","rb") as val_file:
    validation_response = client.files.create(file=val_file,purpose="fine-tune")

training_file_id= training_response.id
validation_file_id=validation_response.id

print("training file id :",training_file_id)
print("validation file id :",validation_file_id)
```

## Step 2: Create and Monitor the Fine-Tuning Job

```python
# Create the fine-tuning job using a supported base model like gpt-4o-mini
try:
    job = client.fine_tuning.jobs.create(
        model="gpt-4.1-mini",
        training_file=training_file_id,
        validation_file=validation_file_id,
        suffix="elements-limerick",
    )
except Exception as exc:
    if getattr(exc, "status_code", None) != 403:
        raise

    print("Fine-tuning job was not created: this organization no longer has access.")
    job = type(
        "UnavailableJob",
        (),
        {"id": "not created", "status": "unavailable"},
    )()

job_id = job.id
print("Fine-tuning Job ID:", job_id)
print("Initial Job Status:", job.status)
```
