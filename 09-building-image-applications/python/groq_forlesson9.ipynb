```python
import base64
import os
from dotenv import load_dotenv
from openai import OpenAI
from PIL import Image 

load_dotenv()

client=OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

print("Open ai client initialized successfully")
```

```python
prompt_text = "Cinematic image of porche car in nework street with cool composition and brad pit at side posing like a model beside it lighthing and mood must be peak aura"
print("generating image.... please wait a moment")

generation_response = client.images.generate(
    model="gpt-image-1",
    prompt=prompt_text,
    size="1024x1024",
    n=1

)

image_dir=os.path.join(os.curdir, "images")

if not os.path.isdir(image_dir):
    os.makedirs(image_dir)

image_path=os.path.join(image_dir,"generated-image.png")


image_data = generation_response.data
if not image_data:
    raise RuntimeError("Image generation returned no image data.")

image_b64 = image_data[0].b64_json
if not image_b64:
    raise RuntimeError("Image generation returned no image content.")

generated_image_bytes=base64.b64decode(image_b64)

with open(image_path,"wb") as image_file:
    image_file.write(generated_image_bytes)

print(f"Success! image saves locally at :{image_path}")    
```

```python
image = Image.open(image_path)
image.show ()

print("Image displayed in your desktop viewer!")

```

```python
disallow_list = "swords, violence, blood, gore, nudity, adult content"

meta_prompt = f"""You are an assistant designer that creates safe family-friendly illustrations.
The image must be appropriate for children, in vibrant color, and in landscape orientation.
Strictly avoid any of the following elements:
{disallow_list}
"""

safe_prompt = f"{meta_prompt}\nCreate an illustration of a historical monument in Paris at sunset with a child looking on."

print("Generating safe, guided image...")

safe_response=client.images.generate(
    model="gpt-image-1",
    prompt=safe_prompt,
    n=1,
    size="1024x1024"


)

safe_image_bytes=base64.b64decode(safe_response.data[0].b64_json)
safe_image_path=os.path.join(image_dir, "safe-monument-image.png")

with open(safe_image_path,"wb") as f:
    f.write(safe_image_bytes)

Image.open(safe_image_path).show()

print("safe image generated and saved successfully!")
```
