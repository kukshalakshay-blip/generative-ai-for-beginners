## Part 2: Multimodal Vision Processing (Mirroring Phi-3.5 Vision)

```python
img=Image.new('RGB',(300,200),color=(73,109,137))
img_path="sample_chart.png"
img.save(img_path)

def encode_image(image_path):
    with open(image_path,'rb') as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')

base64_image=encode_image(img_path)

```

```python
response=client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role":"user",
            "content":[
                {"type":"text","text":"Describe the color and dimensions of this generated chart image."},
                {"type": "image_url", "image_url": {"url": f"data:image/png;base64,{base64_image}"}}
            ]
        }
    ],
    max_tokens=300
)

print("Vision Model Response:")
print(response.choices[0].message.content)
```

## Part 3: Structured Tool Routing & Agentic JSON Output (Mirroring MoE Demo)

```python
sys_msg = """You are a helpful AI assistant capable of using tools like 'Blog' and 'Translate'.
You must always respond in strict JSON format containing an array of steps with 'agentid', 'tool_name', 'input', and 'output' key-value pairs."""

query = "Write something about Generative AI with small models, translate it to Chinese"

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": sys_msg},
        {"role": "user", "content": query}
    ],
    response_format={"type": "json_object"},
    temperature=0.3
)
```
